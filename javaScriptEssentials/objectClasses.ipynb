# Classes and Objects

1. Define objects and classes
2. Create objects and classes
3. Comparison of object literal and function constructors

- Help create and organise code
- Model real-world functionality

A Class is a Blueprint or template for creating object in JavaScript
- it defines **struture** and **behavior**
- Specify instance of a class
- Represents entities with unique values
- Encapsulates state and behavior in a single unit

|Person|Property|Method|
|:--|:--|:--|
|object|`Name`, `age`|`sayHello`|

````typescript
class Person {
    // Constructor to initialize the first name and last name of the person
    constructor(firstName, lastName) {
        this.firstName = firstName; //
        this.lastName = lastName;
    }
    // Method to get the full name of the person
    getFullName() {
        return `${this.firstName} ${this.lastName}`;
    }
}

// Create an instance of the Person class
const person1 = new Person('John', 'Doe');
console.log(person1) //outputs: Person object { firstName: "John", lastName: "Doe" }
console.log(person1.getFullName()); //outputs: John Doe
````

````typescript
class Car {
    // Constructor to initialize the make, model, and year of the car
    constructor(make, model, year) {
        this.make = make;
        this.model = model;
        this.year = year;
    }
    // Method to start the engine
    startEngine() {
        console.log(`The ${this.year} ${this.make} ${this.model}'s engine is running.`);
    }
}
// Create an instance of the Car class
const myCar = new Car('Toyota', 'Camry', 2022); // object literal syntax
myCar.startEngine(); //outputs: The 2022 Toyota Camry's engine is running.
console.log(myCar.startEngine()); //outputs: The 2022 Toyota Camry's engine is running.
````

# Object literals

- Purpose: Create ad hoc object (without defining a **Class**)
- Use case: storing individual customer orders
- Scenario: Unique details of specific instance
- Advantage: creation of one-off objects

````typescript
// object literals
const person = {
    // Properties of the person object
    firstName: 'Alice',
    lastName: 'Johnson',
    // Method to get the full name of the person
    getFullName: function() {
        return `${this.firstName} ${this.lastName}`;
    }
};

console.log(person.getFullName()); //outputs: Alice Johnson
````

# Function constructor
- Constructors are used to initialise objects
- Named with a Capital letter

    - Purpose: Ideal for creating **multiple instances** with *shared* **properties** and **methods**
    - Use case: Creating instance of object like `Car`
    - Scenario: Manage and model multiple similar objects
    - Advantage: Provide a structure and organization
    - Application: Object Oriented Programming (OOP) JavaScript Programming

````typescript
function Kar (make, model) {
    this.make = make;
    this.model = model;
}

const kar1 = new Kar('Toyota', 'Camry');
const kar2 = new Kar('Honda', 'Civic');

console.log("Kar1 details are", kar1); //outputs: Toyota
console.log("Kar2 details are", kar2); //outputs: Honda
````

# Using dot notation

- Access object properties
- Use dot(.) followed by property name
- both bracket and dot notation are the same

````typescript
const person = {
    firstName: "John",
    lastName: "Doe",
    age: 30,
};

// example of accessing object properties using dot notation
console.log(person.firstName); //outputs: John
console.log(person.lastName); //outputs: Doe
console.log(person.age); //outputs: 30
````

# Bracket notation

- Access object properties
- Use when propert name has special charaters or spaces
- use to access dynamic content like variables
- both bracket and dot notation are the same

```js
const propertyName = "firstName";
console.log(person[propertyName]); // output: "John"
```

````typescript
// bracker notation
const person = {
    "first name": "John",
    "last name": "Doe",
    age: 30,
};
console.log(person["first name"]); //outputs: John
console.log(person["last name"]); //outputs: Doe
console.log(person.age); //outputs: 30
````

# Array of objects
- place objects using square brackets
- access elements using array index `[0]`
    - access object property using dot notation `students[0].name`

- iterating through arrays: Common methods
    - For loop
    - ForEach loop

````typescript
const students = [
    { name: "Alice",age: 25 },
    { name: "Bob",age: 22 },
    { name: "Charlie",age: 28 },
];

const studentNames = students.map(student => student.name); 
console.log(studentNames); //outputs: ["Alice", "Bob", "Charlie"]

console.log(students[0].name); //outputs: Alice
console.log(students[2].age); //outputs: 28
````

````typescript
const students = [
    { name: "Alice",age: 25 },
    { name: "Bob",age: 22 },
    { name: "Charlie",age: 28 },
];

for (let i = 0; i < students.length; i++) {
    console.log(students[i].name); //outputs: Alice, Bob, Charlie
}

console.log(students[0].name); //outputs: Alice
console.log(students[2].age); //outputs: 28
````

# forEach

````typescript
const students = [
    { name: "Alice",age: 25 },
    { name: "Bob",age: 22 },
    { name: "Charlie",age: 28 },
];

students.forEach(student => {
    console.log(student.name); //outputs: Alice, Bob, Charlie
});
````

# push() and Pop()

````typescript
const students = [
    { name: "Alice",age: 25 },
    { name: "Bob",age: 22 },
    { name: "Charlie",age: 28 },
];

students.push({ name: "David", age: 30 }); // Add a new student to the array

console.log(students); //
````

````typescript
const students = [
    { name: "Alice",age: 25 },
    { name: "Bob",age: 22 },
    { name: "Charlie",age: 28 },
];

const removedStudent = students.pop(); // Remove the last student from the array
console.log(removedStudent);

console.log(students); 
console.log(students[2].age); //outputs: 28
````

````typescript
const students = [
    { name: "Alice",age: 25 },
    { name: "Bob",age: 22 },
    { name: "Charlie",age: 28 },
];

const adults = students.filter(student => student.age >= 25); // Filter students who are 25 or older
const studentNames = adults.map(student => student.name); // Get the names of the filtered students
console.log(adults); // outputs: [{ name: "Alice", age: 25 }, { name: "Charlie", age: 28 }]
console.log(studentNames); // outputs: ["Alice", "Charlie"]
````

````typescript
const students = [
    { name: "Alice",age: 25 },
    { name: "Bob",age: 22 },
    { name: "Charlie",age: 28 },
];

const student = students.find(student => student.name === "Bob"); // Find the student with the name "Bob"
console.log(student.age); // outputs: 22
````

````typescript
const text = "Hello World";

const lowercaseText = text.toLowerCase(); // "hello world"
const uppercaseText = text.toUpperCase(); // "HELLO WORLD"

console.log('The lowercase for text is', lowercaseText); // outputs: hello world
console.log('The uppercase for text is', uppercaseText); // outputs: HELLO WORLD
````

````typescript
// string search methods
const sentence = "The quick brown fox jumps over the lazy dog.";
const hasFox = sentence.includes("fox"); // true
const indexOfDog = sentence.indexOf("fox"); // 16

console.log(hasFox); // outputs: true
console.log(indexOfDog); // outputs: 16
````

````typescript
// substring extracting
const text = "Hello, World!";
const subText1 = text.substring(0, 5); // "Hello"
const subText2 = text.substring(7); // "World!"
const subText3 = text.substring(7, 5); // "World"
console.log(subText1); // outputs: Hello
console.log(subText2); // outputs: World!
console.log(subText3); // outputs: World
````

````typescript
// replace substring methods
// substring extracting
const text = "Hello, World!";
const updatedText = text.replace("Hello", "Universe"); // "Universe, World!"
console.log(updatedText); // outputs: Universe, World!
````

````typescript
let count = 0;

const intervalId = setInterval(function() {
    console.log("Count: " + count);
    count++; //1 increments by 1
    if (count > 5) {
        clearInterval(intervalId); // Stop after 6 iterations
    }
}, 1000); // Displayed every second
````
