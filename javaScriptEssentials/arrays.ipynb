// JavaScript for booking seats
// "X" = booked, "O" = available
const theaterSeats = [
    ['X', 'O', 'X'],
    ['O', 'X', 'O'],
    ['X', 'O', 'X']
];

function bookSeat(row, col) {
    if (theaterSeats [row][col] === 'O'){
        theaterSeats [row][col] = 'X';
        updateSeatStatus (row, col, 'booked');
        alert(`Seat ${String.formCharCode(65 + row)} ${col + 1} is booked.`)
    } else {
        alert(`Seat ${String.formCharCode(65 + row)} ${col + 1} is already taken.`);
    }
}

function updateSeatStatus(row, col, status){
    const seats = document.getElementByClassName('seat');
    const index = row * 3 + col;
    seats[index].classList.remove('available', 'booked');
    seats[index].classList.add(status);
}

function bookRandomSeat() {
    const availableSeats = [];

}

## The methods
|Array Methods|Description|
|:--|:--|
|push()|add 1 or more elements to the end of an array|
|pop()|removes the last element from an array and returns that element|
|shift()|removes the first element from an array and returns that element|
|unshift()|add 1 or more elements to the beginning of an array|
|splice()|changes conten of array, removes, replaces, adds elements|

````typescript
/* add 1 or more elements to the end of an array */
const fruits = ["apple", "banana"];
fruits.push("orange", "strawberry");
console.log(fruits); // Output: ["apple", "banana", "orange", "strawberry"]
````

````typescript
/* removes the last element from an array and returns that element */
const  fruits = ["apple", "banana", "orange"];
const removedFruit = fruits.pop(); //store the removed element
console.log(removedFruit); // Output: "orange"
console.log("removed fruits is", removedFruit); // Output: "removed fruits is orange"
````

````typescript
/* removes the last element from an array and returns that element */
const  fruits = ["apple", "banana", "orange"];
const removedFruit = fruits.shift(); //store the removed element
console.log(removedFruit); // Output: "apple"
console.log("removed fruits is", removedFruit); // Output: "removed fruits is apple"
````

````typescript
/* add 1 or more elements to the beginning of an array and returns a new array*/
const fruits = ["apple", "banana"];
fruits.unshift("orange", "strawberry");
console.log(fruits); // Output: ["orange", "strawberry", "apple", "banana"]
````
