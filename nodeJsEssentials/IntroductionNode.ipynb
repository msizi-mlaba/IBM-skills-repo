# http/createServer

http package is used for establishing remote connections to a server or to create a server which listens to client. createServer - Takes a requestListener, a function which takes request and response parameters, where request is the handle to the request from the client and response is the handle to be sent to the client.

```typescript
const http = require('http');

const requestListener = function(req, res) {
    res.writeHead(200); res.end('Hello, World!');
}

const port = 8080;const server = http.createServer(requestListener);
console.log('server listening on port: '+ port);
server.listen(port);
```

# Asynchronous I/O with Callback Programming

- Network operations run in an asynchronous () manner
    - Example, the **response** from a web service call **might not return immediately**
- When and application blocks (waits) for a network operation to complete, that application wastes processing time on the server
- Node.js makes all network operations in a non-blocking manner
- Network operations return immediately
- to Handle the result from a network call, you can write a callback function that node calls when the network operation completes
- non-blocked operations return immediately without additional processing time on the server/

|Application|callback function|Node framework|remote server|
|:--|:--|:--|:--|
|1.call http.request|--->|2.send HTTP resquest message to web service|--->|
|3.Return from http.request() call|<---|4.Recieve HTTP response message|<---|
|<--|---|5.node.js starts callback to handle web service response||


```typescript
//sending Http resquest
// 
let options = {
    host: 'w1.weather.gov'
    path: '/xml/current_obs/KSFO.xml'
};

http.request(options,
// anynomous callback function with 1 parameter, a response object
function(response) {
    let buffer="";
    let result="";
// two events, `data` event and `end` event
    response.on('data', function(chunk) {
        buffer += chuck;
    }) // data event also has anynomous callback function

    response.on('end', function() {
        console.log(buffer);
    })
}), end();


//http.request(options, [callback function]);

// define the host authetication, ports, authentication, protocol 
```

# Promise use case

Description: Promises are used when the processing time of the function we invoke takes time like remote URL access, I/O operations file reading, etc.

```typescript
let prompt = require('prompt-sync')();
let fs = require('fs');

const methCall = new Promise((resolve,reject)=>{
 let filename = prompt('What is the name of the file ?');
 try {
  const data = fs.readFileSync(filename, {encoding:'utf8', flag:'r'});
  resolve(data);
 } catch(err) {
  reject(err)
 }
});

console.log(methCall);

methCall.then(
 (data) => console.log(data),
 (err) => console.log("Error reading file")
);
```

```typescript
import { rejects } from "node:assert";
import { resolve } from "node:dns";

// creating a new Promise object and assinging it to the variable myPromise
const myPromise = new Promise((resolve, reject)=>{
    //Stimulating a condition with a boolean variable 'success'
    let success = true;
    // if the condition is true, call resolve to mark the promise as fulfilled
    if(success){
        resolve('The operation was successful!');
    } else {
        // If the condition is false, call reject to mark the promise as rejected
        reject('The operation failed');
    }
});
```

### Using promises with .then() and .catch()

```typescript
import { error } from "node:console";

// Execute the Promise and handle the fulfiled and rejected states
myPromise

// Handle the resolved state of the promise
.then((message)=>{
    // This block will execute if the promise is resolved
    console.log(message); // output: 'The operation was successful!'
})

// Handle the rejected state of the promise
.catch((error) => {
    //This block will execute if the promise is rejected
    console.error(error); // output: 'The operation failed'
});

```

### Reading a file with promises

```typescript
// import the 'fs' module and use its promise-based methods
const fs = require('fs').promises;

//Read the content of the file 'example.txt' with 'utf8' encoding
fs.readFile('example.txt', 'utf8')

// Handle the resolved state of the promise
.then((data) => {
    // this block will execute if the file is read successfully
    console.log(data); // Print the file content to the console
})

// Handle the rejected state of the promise
.catch((err) => {
    // this block will execute if there is an error reading the file
    console.error('Error reading this file:', err); // Prints the error message to the console
})
```

### Using async/await function

Benefits
- Readability
- Familiar error handling
- easier debugging



```typescript
//Async function that wraps the operation
async function myAsyncFunction() {
    // simulate a condition with a boolean variable 'success'
    let success = true;

    // if condition is true, resolve with a success message
    if (success){
        return 'The operation was successful!';
    } else {
        // if condition is false, throw and error message to simulate rejection
        throw new Error('The operation failed!');
    }
}

// Using async function to handle promise
async function executeAsyncFunction (){
    try{
        // Await the sync function call to get the result 
        const result = await myAsyncFunction();
        console.log (result); // output the result if successful
    } catch (error) {
        console.error(error.message) // handle and output any errors
    }
}

// call the async function to execute
executeAsyncFunction();
```

### Using Axios to make GET Requests

```typescript
// Import the axios library
const axios = require('axios');

//Using the axios.get method to make a GET request to the specific URL
axios.get('https://api.example.com/data')

// if the request is successful, the `.then` block is executed.
.then(response => {
    // The response object contains the data returned from the server.
    // We log the `data` property of the response to the console.
    console.log(response.data);
})

// if there is an error during the request, the .catch() block is executed.
.catch(error => {
    // we log an error message to the console along with error object
    // This helps with the debuggin and understanding of what went wrong with the request
    console.log('Error fecthing data', error)
});
```

### Using Axios to make POST request

- Import Axios for HTTP request
- create user data fot POST request
- Use axios.post to send data to server
- Log response data on success
- log error message on failure

```typescript
// import the Axios library.

const axios = require('axios');

// Data to be sent to the POST request. This is a JavaScript object contianing user information.
const data = {
    name: 'John Doe',
    age: 30
};

// using the axios.post method to make a POST request to the specified URL with the data object.
axios.post('https://api.example.com/users', data)

// if the request is successful, the .then() blocked is executed.
.then(response => {
    // we log a data message along with a `data` property to the console
    console.log('user created:', response.data);
})

// if the request is an error, the .catch block is executed.
.catch(error => {
    //we log a message to the console along with the error
    console.log('Error fecthing user', error)
})
```
